```python
import numpy as np
import matplotlib.pyplot as plt
import os


import seaborn as sbn
sbn.set(context='paper', font_scale=0.8, color_codes=True, palette='deep', style='ticks',
        rc={'mathtext.fontset': 'cm', 'axes.linewidth': 1, 'figure.dpi':300, 'figure.figsize':(2,1.5)})
```

```python
rates = np.logspace(-5,-3,3)
NList = [6, 8, 10, 12]
```

```python
neg = {}
mutualInf = {}
purity = {}
Nsteps = 150

for N in NList:
    neg[N] = np.zeros([len(rates),Nsteps])
    mutualInf[N] = np.zeros([len(rates),Nsteps])
    purity[N] = np.zeros([len(rates),Nsteps])
    for (ind,r) in enumerate(rates):
        
        homepath = "RC_mobius/N=" + str(N) + "/r=" + str(r)  + "p=0.13/"
        Nfiles = len(os.listdir(homepath))//3
        for fi in range(Nfiles):
            x = np.loadtxt(homepath + "negativity_" + str(fi) + ".out","a")
            neg[N][ind,:] += np.array(x).astype(float)/Nfiles

            y = np.loadtxt(homepath + "I_" + str(fi) + ".out","a")
            mutualInf[N][ind,:] += np.array(y).astype(float)/Nfiles

            z = np.loadtxt(homepath + "purity_" + str(fi) + ".out","a")
            purity[N][ind,:] += np.array(z).astype(float)/Nfiles
            
```

```python
r = 0
plt.figure()
for N in (NList):
    plt.plot(neg[N][r,:])
plt.legend(NList)

plt.figure()
for N in (NList):
    plt.plot(mutualInf[N][r,:])
plt.legend(NList)


plt.figure()
for N in (NList):
    plt.plot(purity[N][r,:])
```
